## **Wireless Channel Impairments**

### **1. Large Scale Path Loss and Shadowing**  
  
The large scale path loss and shadowing can be modeled as:
  
$P_r(d) = P_r(d0) −10nlog_{10}(d/d_0) + X_σ$  
  
Where:
- $P_r(d)$: Received power at distance $d$
- $P_r(d0)$: Received power at reference distance $d_0$
- $n$: Path loss exponent
- $X_σ$: Shadowing effect (normally distributed random variable)

```python
import numpy as np
from numpy import log10, logspace
import matplotlib.pyplot as plt

# parameters    
d0 = 1                   # reference distance (m)
d = logspace(0, 3, 1000) # distance range: 1 to 1000m
n = 3.2                  # path loss exponent
sigma = 6                # shadowing standard deviation (dB)
pr0 = -30                # received power at reference distance (dBm)

# path loss model

# Calculate the mean received power at distance d
mean_pr = pr0 - 10 * n * log10(d / d0)
    
# Add shadowing effect (Gaussian noise)
shadowing = np.random.normal(0, sigma, size=d.shape)
    
# Total received power
pr = mean_pr + shadowing

# 1. Open the figure window
plt.figure(figsize=(8, 5)) # Optional: sets a clean aspect ratio

# 2. Plot the smooth deterministic line
plt.semilogx(d, mean_pr, color='blue', linewidth=1.5, label='Mean Path Loss')

# 3. Use scatter for the random shadowing points to force the exact MATLAB look
# 's=10' controls the dot size. We set the x-axis to log scale manually right after.
plt.scatter(d, pr, color=(0.8, 0.2, 0.2), s=10, marker='.', label='With Shadowing')
plt.xscale('log') 

# 4. Formatting to match MATLAB exactly
plt.xlabel('Distance (m)')
plt.ylabel('Received Power (dBm)')
plt.title('Large-Scale Path Loss with Shadowing')
plt.legend(loc='best')

# 5. Render
plt.show()
```
